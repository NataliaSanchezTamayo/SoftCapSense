```python
import pandas as pd
from sklearn.metrics import confusion_matrix, accuracy_score
import numpy as np

# Note: In paper, Table 1 ("Performance comparison of cross-sensor generalization and transfer learning results"), Sensor 2 is reported as Sensor 1.
```

## LOGO-CV results for model trained on S2,S3,S4 and tested on S1

```python
file_path = "../../Data/Results/LOGO-CV_Single_Point_Contact_Estimation_Results/1. Train_s2s3s4_Test_s1/test_error_results_s1.xlsx"
test_error_dataset_s1 = pd.read_excel(file_path)
```

```python
test_error_dataset_s1[[8,9,10]] = (test_error_dataset_s1.iloc[:,:3].sub(test_error_dataset_s1.iloc[:,4:7].values)).abs()
test_error_dataset_s1_filtered = test_error_dataset_s1
test_error_dataset_s1_filtered = test_error_dataset_s1_filtered[test_error_dataset_s1_filtered.iloc[:,0]>= -17.25] 
test_error_dataset_s1_filtered = test_error_dataset_s1_filtered[test_error_dataset_s1_filtered.iloc[:,0]<= -0.75]
test_error_dataset_s1_filtered = test_error_dataset_s1_filtered[test_error_dataset_s1_filtered.iloc[:,1]>= -17.25] 
test_error_dataset_s1_filtered = test_error_dataset_s1_filtered[test_error_dataset_s1_filtered.iloc[:,1]<= -0.75]
```

```python
y_true = test_error_dataset_s1_filtered.iloc[:,3].values
y_pred = test_error_dataset_s1_filtered.iloc[:,7].values

# Threshold the predicted values at 0.5
y_pred_binary = (y_pred >= 0.5).astype(int)

# Generate the confusion matrix using sklearn
cm = confusion_matrix(y_true, y_pred_binary)
print("Confusion Matrix:")
print(cm)

# Calculate classification accuracy
accuracy_s1 = accuracy_score(y_true, y_pred_binary)
print(f"Classification Accuracy: {accuracy_s1:.3f}")
```

```python
# Force Filtering
test_error_dataset_s1_filtered = test_error_dataset_s1_filtered[test_error_dataset_s1_filtered.iloc[:,3]==1] 
test_error_dataset_s1_filtered.iloc[:,2] = test_error_dataset_s1_filtered.iloc[:,2]*-1
test_error_dataset_s1_filtered = test_error_dataset_s1_filtered[test_error_dataset_s1_filtered.iloc[:,2]<=2] 
```

```python
# Sort the DataFrame by the X Target and Y Target columns in descending order
test_error_dataset_force_filtered_sorted = test_error_dataset_s1_filtered.sort_values(
    by=[test_error_dataset_s1_filtered.columns[0], test_error_dataset_s1_filtered.columns[1]],
    ascending=[False, False]
).reset_index(drop=True)

# Initialize an index array to mark rows for deletion
rows_to_delete = np.zeros(len(test_error_dataset_force_filtered_sorted), dtype=bool)

# Loop through the DataFrame to check for consecutive equal values in the X Target and Y Target columns
for i in range(len(test_error_dataset_force_filtered_sorted) - 2):
    # Check for 3 consecutive equal values in X Target
    if not (
        test_error_dataset_force_filtered_sorted.iloc[i, 0] == test_error_dataset_force_filtered_sorted.iloc[i + 1, 0] == test_error_dataset_force_filtered_sorted.iloc[i + 2, 0]
    ):
        rows_to_delete[i:i + 3] = True  # Mark the rows for deletion

    # Check for 3 consecutive equal values in Y Target
    if not (
        test_error_dataset_force_filtered_sorted.iloc[i, 1] == test_error_dataset_force_filtered_sorted.iloc[i + 1, 1] == test_error_dataset_force_filtered_sorted.iloc[i + 2, 1]
    ):
        rows_to_delete[i:i + 3] = True  # Mark the rows for deletion

# Delete the marked rows
test_error_dataset_s1_filtered_cleaned = test_error_dataset_force_filtered_sorted[~rows_to_delete]
```

```python
# Optionally, display the pre-cleaned DataFrame
print("Size of test_error_dataset_s1_filtered:", test_error_dataset_s1_filtered.shape)
# Optionally, display the cleaned DataFrame
print("Size of test_error_dataset_s1_filtered_cleaned:", test_error_dataset_s1_filtered_cleaned.shape)
```

```python
# Calculate mean of 9th, 10th, and 11th columns
mean_x_error = test_error_dataset_s1_filtered_cleaned.iloc[:, 8].mean()
mean_y_error = test_error_dataset_s1_filtered_cleaned.iloc[:, 9].mean()
mean_force_error = test_error_dataset_s1_filtered_cleaned.iloc[:, 10].mean()
mean_sqrt_loc_val = np.sqrt(mean_x_error ** 2 + mean_y_error ** 2)

print("Mean of X Error (mm):", mean_x_error)
print("Mean of Y Error (mm):", mean_y_error)
print("Mean of Force Error (N):", mean_force_error)

# Previous calculation of resultant MAE for contact location 
print("Mean of sqrt loc (mm):",mean_sqrt_loc_val)
```

```python
abs_x = test_error_dataset_s1_filtered_cleaned.iloc[:, 8]
abs_y = test_error_dataset_s1_filtered_cleaned.iloc[:, 9]
abs_force = test_error_dataset_s1_filtered_cleaned.iloc[:, 10]

# 1. MEE (Mean Euclidean Error) - Spatial
individual_distances_s1 = np.sqrt(abs_x**2 + abs_y**2)
mee_val_s1 = individual_distances_s1.mean()

# 3. MAE (Mean Absolute Error)
mae_x_s1 = abs_x.mean()
mae_y_s1 = abs_y.mean()
mae_force_s1 = abs_force.mean()

# 4. L2 Norm of MAEx and MAEy (Spatial Bias)
l2_norm_mae_s1 = np.sqrt(mae_x_s1**2 + mae_y_s1**2)

print(f"MEE: {mee_val_s1:.4f} mm")
print(f"MAE X: {mae_x_s1:.4f} mm, MAE Y: {mae_y_s1:.4f} mm")
print(f"L2 Norm of MAE: {l2_norm_mae_s1:.4f} mm")
print("-" * 30)
print(f"MAE Force: {mae_force_s1:.4f} N")
```

## Calculating average and standard deviation across the results

```python
combined_results = {
    "Sensor (Left-Out)": ["Sensor 1", "Sensor 2", "Sensor 3", "Sensor 4"],
    "Training Sensors": ["Sensor 2, 3, 4", "Sensor 1, 3, 4", "Sensor 1, 2, 4", "Sensor 1, 2, 3"],
    "Location Error [mm]": [mee_val_s1, mee_val_s2, mee_val_s3, mee_val_s4],
    "Force Error [N]": [mae_force_s1, mae_force_s2, mae_force_s3, mae_force_s4],
    "Classification Accuracy [%]": [accuracy_s1, accuracy_s2, accuracy_s3, accuracy_s4]
}

# Convert to DataFrame
combined_results_df = pd.DataFrame(combined_results)

# Compute the average (mean)
average_values = combined_results_df.iloc[:, 2:].mean()

# Compute the standard deviation
std_values = combined_results_df.iloc[:, 2:].std()

# Add to DataFrame
combined_results_df.loc["Average"] = ["", ""] + list(average_values)
combined_results_df.loc["Standard Deviation"] = ["", ""] + list(std_values)

# Print final DataFrame
print(combined_results_df)
```
